```python

# NAME: rentalcar.py

# IMPORT STATEMENTS
# None

# FUNCTION DEFINITIONS
# Define verify_car_class
def verify_car_class():
  while True:
    user_car_class = input("Please enter car classification (B, S, or L): ").upper()
   # Create a while loop to ensure user enters a valid car classification (B, S, L)
    if user_car_class in ("B", "S", "L"):
      return user_car_class
    else:
      print("Invalid classification, try again!")
   # Program will proceed until a valid entry is entered by the user

# Define a function: total_miles with parameters of starting mileage (start) and ending mileage (end)
def total_miles(start_mileage, ending_mileage):
  # subtract start from ending mileage to find total miles

  total = (ending_mileage - start_mileage)

# from total miles subtract one hundred to get billed_miles (first 100 miles free)
  billed_miles = total - 100

# if driven less than 100 miles, it's free
  if billed_miles < 0:
    billed_miles = 0

  return billed_miles
def rentalcar_py(user_car_class, billed_miles, rental_days):
  if user_car_class == "B":
  # if car_class == "B" :
  # rate is $40/day
    rate = 40
 # elif car_class == "S":
    # rate is $60/day
  elif user_car_class == "S":
    rate = 60
  # else car_class == "L":
    # rate is $80/day
  else:
    rate = 80

  # rental_day_payment = rental_days * rate
  rental_day_payment = rental_days * rate
  # mile_payment = billed_miles * $0.35
  mile_payment = billed_miles * 0.35
  # total_bill = mile_payment + rental_day_payment
  total_bill = mile_payment + rental_day_payment
  # return total_bill
  return total_bill

# Define what each car classification stands for
def full_car_name(user_car_class):
  # B = Basic
  if user_car_class == "B":
    return "Basic"
  # S = Standard
  elif user_car_class == "S":
    return "Standard"
  # L = Luxury
  else:
    return "Luxury"
# PROGRAM LOGIC

# Ask user for input of car classification: (B, S, or L)
# Decided to include in the while loop

user_car_class = verify_car_class()

# Ask user for mile number at the start of their trip

start_mileage = float(input("Please enter starting mileage: "))

# Ask user for mile number at the end of their trip

ending_mileage = float(input("Please enter ending mileage: "))


# Ask user the number of days they rented their car

rental_days = int(input("Please enter number of days rented: "))

# Call total_miles function by defining it terms of billed_miles

billed_miles = total_miles(start_mileage, ending_mileage)

# Call full_car_name

car_name = full_car_name(user_car_class)

# Call rentalcar.py and return their rental summary (car classification, total miles driven, miles charged, and total bill)

total_bill = rentalcar_py(user_car_class, billed_miles, rental_days)
print("---RENTAL SUMMARY----")
print("Classification: " + car_name)
print("Total miles: " + str(ending_mileage - start_mileage))
print("Billed miles: " + str(billed_miles))
print("Total bill: $" + str(total_bill))


```
