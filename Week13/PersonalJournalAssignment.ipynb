```python
from os import write
# NAME: journal.py
# IMPORT STATEMENTS #



# FUNCTION DEFINITIONS #

# Define a function called write_entry that uses a filenane and a string (journal entry) as a parameter
def write_entry(journal_entries, journal_entry):
  # Open journal file with the append ("a") mode
  with open("journal_entries.txt", "a") as journal_entries:
  # Write entry and make sure each entry is a newline (\n)
    journal_entries.write(journal_entry + "\n")
  # Close file when finished
  journal_entries.close()

# Define a function called read_journal that accepts the journal filename as a parameter
def read_journal(filename):
  try:
  # Open the file using the read ("r") mode
    with open(filename, "r") as journal_entries:
    # Read the journal file
      contents = journal_entries.read()
      if contents == "":
        print("Sorry, no entries yet!")
      else:
    # Print contents of the file (if there are any)
        print(contents)
  except FileNotFoundError:
    print("No journal found. Start by writing an entry!")


# PROGRAM LOGIC #
# Create a While loop
while True:
# Ask user if they want to write a new entry or read their journal
  user_choice = input("Would you like to (w)rite or (r)ead in your journal? (Type quit to exit) ")
  user_choice = user_choice.strip().lower()
   # If user responds write ("W"),
  if user_choice == "w":
     # Ask for their thoughts
    journal_entry = input("What would you like to write today? ")
     # Call the write_entry function
    write_entry("journal_entries.txt", journal_entry)
    # Elif they choose to read ("r"),
  elif user_choice == "r":
      # Call the read_journal function
      read_journal("journal_entries.txt")
    # Elif they type "quit"
  elif user_choice == "quit":
      # Break
      break
    # Else
  else:
      # Print ("Invalid entry, try again!")
      print("Invalid entry, try again!")
```
