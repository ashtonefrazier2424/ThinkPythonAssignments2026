```python
# NAME: rectangle.py

# IMPORT STATEMENTS #
# NONE #


# FUNCTION DEFINITIONS #
# Define a function calculate_rectangle with parameters: length and width
def calculate_rectangle(length, width):
  # Calculate the area using the provided length and width (l * w)
  area = length * width
  # Calculate the perimeter using provided length and width (2*(l + w))
  perimeter = 2 * (length + width)
  # Return length AND width as a tuple (area, perimeter)
  return (area, perimeter)



# PROGRAM LOGIC #

# Ask the user to input a length (*make it a float*)
length = float(input("Please enter the rectangle's length: "))

# Ask user for a width (*make a float*)

width = float(input("Please enter the rectangle's width: "))

# Call calculate_rectangle using user's input

print(calculate_rectangle(length, width))

# Unpack the returned tuple (area, perimeter) into two separate variables (my_area and my_perimeter)

# Print the calculation results
```
