First, let's create a placeholder list to store your expenses. We'll add this at the top of your program logic section.

```python
expenses = [] # This list will hold all your expense dictionaries
```

Now, let's define a function that can filter the `expenses` list based on a given key (like 'category' or 'vendor') and a value.

```python
def filter_expenses(expense_list, filter_key, filter_value):
    """Filters a list of expenses by a given key and value."""
    filtered = []
    for expense in expense_list:
        if filter_key in expense and str(expense[filter_key]).lower() == str(filter_value).lower():
            filtered.append(expense)
    return filtered

def display_expenses(expense_list):
    if not expense_list:
        print("No expenses to display.")
        return
    print("\n--- Your Expenses ---")
    for i, exp in enumerate(expense_list):
        print(f"{i+1}. Category: {exp.get('category', 'N/A')}, Vendor: {exp.get('vendor', 'N/A')}, Amount: ${exp.get('amount', 0):.2f}, Date: {exp.get('date', 'N/A')}")
    print("---------------------")
```

Next, we'll modify your menu loop to allow adding expenses and then filtering/viewing them. We'll replace the existing `PROGRAM LOGIC` with an updated version.

```python
import datetime # For getting the current date

# PROGRAM LOGIC #
while True:
    print("\n--- Expense Tracker Menu ---")
    print("1. Add new expense")
    print("2. View all expenses")
    print("3. Get total expenses")
    print("4. Exit")

    choice = input("Enter your choice (1-4): ")

    if choice == '1':
        print("\n--- Add New Expense ---")
        category = input("Enter expense category (e.g., Food, Transport): ")
        vendor = input("Enter vendor: ")
        while True:
            try:
                amount = float(input("Enter amount: "))
                break
            except ValueError:
                print("Invalid amount. Please enter a number.")
        date_str = input(f"Enter date (YYYY-MM-DD, default: {datetime.date.today()}): ")
        if not date_str:
            date_str = str(datetime.date.today())

        expense = {
            'category': category,
            'vendor': vendor,
            'amount': amount,
            'date': date_str
        }
        expenses.append(expense)
        print("Expense added successfully!")

    elif choice == '2':
        print("\n--- View Expenses ---")
        filter_option = input("Filter by (category/vendor/none)? ").lower()
        if filter_option == 'category':
            filter_value = input("Enter category to filter by: ")
            filtered_list = filter_expenses(expenses, 'category', filter_value)
            display_expenses(filtered_list)
        elif filter_option == 'vendor':
            filter_value = input("Enter vendor to filter by: ")
            filtered_list = filter_expenses(expenses, 'vendor', filter_value)
            display_expenses(filtered_list)
        elif filter_option == 'none':
            display_expenses(expenses) # Display all expenses
        else:
            print("Invalid filter option. Displaying all expenses.")
            display_expenses(expenses)

    elif choice == '3':
        print("\n--- Get Total Expenses ---")
        # This part can be expanded to sum based on filters too
        total = sum(exp.get('amount', 0) for exp in expenses)
        print(f"Total expenses: ${total:.2f}")

    elif choice == '4':
        print("Exiting Expense Tracker. Goodbye!")
        break # Exit the loop
    else:
        print("Invalid choice. Please enter a number between 1 and 4.")
```
