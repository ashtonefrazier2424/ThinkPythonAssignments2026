## Generate Initial Code for Book Tracker

### Subtask:
Generate the Python code that includes a `Book` class, functions for adding, updating, and listing books, as well as saving and loading data to/from a JSON file. The code will also include a simple command-line interface to interact with these functionalities.


**Reasoning**:
The subtask requires generating a complete Python script for a book tracker with a `Book` class, CRUD operations for books, JSON persistence, and a CLI. I will provide a single code block that implements all these requirements.



```python
import json
import os

class Book:
    def __init__(self, title, author, isbn=None, read_status='unread'):
        if not title or not author:
            raise ValueError("Title and author cannot be empty.")
        self.title = title
        self.author = author
        self.isbn = isbn
        self.read_status = read_status

    def to_dict(self):
        return {
            'title': self.title,
            'author': self.author,
            'isbn': self.isbn,
            'read_status': self.read_status
        }

    @classmethod
    def from_dict(cls, data):
        return cls(data['title'], data['author'], data.get('isbn'), data.get('read_status', 'unread'))

    def __str__(self):
        return f"Title: {self.title}, Author: {self.author}, ISBN: {self.isbn if self.isbn else 'N/A'}, Status: {self.read_status}"


def add_book(books, title, author, isbn=None, read_status='unread'):
    for book in books:
        if book.isbn and book.isbn == isbn:
            print("Error: A book with this ISBN already exists.")
            return books

    try:
        new_book = Book(title, author, isbn, read_status)
        books.append(new_book)
        print(f"Book '{title}' added successfully.")
    except ValueError as e:
        print(f"Error adding book: {e}")
    return books


def update_book(books, isbn, new_title=None, new_author=None, new_read_status=None):
    found = False
    for book in books:
        if book.isbn == isbn:
            if new_title: book.title = new_title
            if new_author: book.author = new_author
            if new_read_status: book.read_status = new_read_status
            found = True
            print(f"Book with ISBN {isbn} updated successfully.")
            break
    if not found:
        print(f"Error: Book with ISBN {isbn} not found.")
    return books


def list_books(books, status_filter=None):
    if not books:
        print("No books in the collection.")
        return

    filtered_books = []
    if status_filter:
        filtered_books = [book for book in books if book.read_status == status_filter]
        if not filtered_books:
            print(f"No books found with status: {status_filter}")
            return
    else:
        filtered_books = books

    print("\n--- Book List ---")
    for book in filtered_books:
        print(book)
    print("-------------------")


def save_books(books, filename='books.json'):
    try:
        with open(filename, 'w') as f:
            json.dump([book.to_dict() for book in books], f, indent=4)
        print(f"Books saved to {filename} successfully.")
    except IOError as e:
        print(f"Error saving books: {e}")


def load_books(filename='books.json'):
    books = []
    if os.path.exists(filename):
        try:
            with open(filename, 'r') as f:
                data = json.load(f)
                books = [Book.from_dict(item) for item in data]
            print(f"Books loaded from {filename} successfully.")
        except json.JSONDecodeError as e:
            print(f"Error decoding JSON from {filename}: {e}. Starting with an empty collection.")
        except IOError as e:
            print(f"Error loading books: {e}. Starting with an empty collection.")
    else:
        print(f"File {filename} not found. Starting with an empty collection.")
    return books


def main():
    books = load_books()

    while True:
        print("\nBook Tracker Menu:")
        print("1. Add a new book")
        print("2. Update a book")
        print("3. List all books")
        print("4. List books by status")
        print("5. Save books")
        print("6. Load books")
        print("7. Exit")

        choice = input("Enter your choice: ")

        if choice == '1':
            title = input("Enter title: ")
            author = input("Enter author: ")
            isbn = input("Enter ISBN (optional): ") or None
            read_status = input("Enter read status (read/unread, default unread): ") or 'unread'
            books = add_book(books, title, author, isbn, read_status)
        elif choice == '2':
            isbn = input("Enter ISBN of the book to update: ")
            new_title = input("Enter new title (leave blank to keep current): ") or None
            new_author = input("Enter new author (leave blank to keep current): ") or None
            new_read_status = input("Enter new read status (read/unread, leave blank to keep current): ") or None
            books = update_book(books, isbn, new_title, new_author, new_read_status)
        elif choice == '3':
            list_books(books)
        elif choice == '4':
            status = input("Enter status to filter by (read/unread): ")
            list_books(books, status)
        elif choice == '5':
            save_books(books)
        elif choice == '6':
            books = load_books()
        elif choice == '7':
            print("Exiting Book Tracker. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

if __name__ == '__main__':
    main()
```
