```python
tasks = []

def add_task():
    task_name = input("Enter task name: ")
    while True:
        priority_input = input("Enter priority (1 for High, 2 for Medium, 3 for Low): ")
        if priority_input in ['1', '2', '3']:
            priority = int(priority_input)
            break
        else:
            print("Invalid priority. Please enter 1, 2, or 3.")

    priority_map = {1: 'High', 2: 'Medium', 3: 'Low'}
    tasks.append({"name": task_name, "priority": priority_map[priority]})
    print(f"Task '{task_name}' added with {priority_map[priority]} priority.")

def view_tasks():
    if not tasks:
        print("No tasks added yet.")
        return

    print("\n--- Your Tasks ---")
    # Sort tasks by priority for better readability (optional)
    sorted_tasks = sorted(tasks, key=lambda x: {'High': 1, 'Medium': 2, 'Low': 3}[x['priority']])
    for i, task in enumerate(sorted_tasks):
        print(f"{i + 1}. Name: {task['name']}, Priority: {task['priority']}")
    print("-------------------\n")

def main():
    while True:
        print("\nPersonal Task Manager Menu:")
        print("1. Add a new task")
        print("2. View all tasks")
        print("3. Exit")

        choice = input("Enter your choice (1-3): ")

        if choice == '1':
            add_task()
        elif choice == '2':
            view_tasks()
        elif choice == '3':
            print("Exiting Task Manager. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 3.")

if __name__ == "__main__":
    main()
```
