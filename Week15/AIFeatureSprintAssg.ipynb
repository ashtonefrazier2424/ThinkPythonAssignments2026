```python
import datetime

tasks = []
COMPLETED_TASKS_FILE = 'completed_tasks.txt'

def add_task():
    task_name = input("Enter task name: ")
    while True:
        priority_input = input("Enter priority (1 for High, 2 for Medium, 3 for Low): ")
        if priority_input in ['1', '2', '3']:
            priority = int(priority_input)
            break
        else:
            print("Invalid priority. Please enter 1, 2, or 3.")

    priority_map = {1: 'High', 2: 'Medium', 3: 'Low'}
    tasks.append({"name": task_name, "priority": priority_map[priority]})
    print(f"Task '{task_name}' added with {priority_map[priority]} priority.")

def view_tasks():
    if not tasks:
        print("No tasks added yet.")
        return

    print("\n--- Your Tasks ---")
    # Sort tasks by priority for better readability
    sorted_tasks = sorted(tasks, key=lambda x: {'High': 1, 'Medium': 2, 'Low': 3}[x['priority']])
    for i, task in enumerate(sorted_tasks):
        print(f"{i + 1}. Name: {task['name']}, Priority: {task['priority']}")
    print("-------------------\n")
    return sorted_tasks # Return sorted tasks for easier completion

def complete_task():
    if not tasks:
        print("No tasks to complete.")
        return

    print("\n--- Complete a Task ---")
    sorted_tasks = view_tasks() # Display tasks and get the sorted list

    while True:
        try:
            task_index_str = input(f"Enter the number of the task to complete (1-{len(sorted_tasks)}): ")
            task_index = int(task_index_str) - 1 # Adjust for 0-based indexing

            if 0 <= task_index < len(sorted_tasks):
                completed_task = sorted_tasks[task_index]
                # Remove the task from the original 'tasks' list
                # Find the actual task object to remove it correctly
                for i, task in enumerate(tasks):
                    if task['name'] == completed_task['name'] and task['priority'] == completed_task['priority']:
                        del tasks[i]
                        break

                with open(COMPLETED_TASKS_FILE, 'a') as f:
                    f.write(f"{datetime.datetime.now()}: {completed_task['name']} (Priority: {completed_task['priority']})\n")
                print(f"Task '{completed_task['name']}' marked as complete and recorded.")
                break
            else:
                print("Invalid task number. Please try again.")
        except ValueError:
            print("Invalid input. Please enter a number.")

def main():
    while True:
        print("\nPersonal Task Manager Menu:")
        print("1. Add a new task")
        print("2. View all tasks")
        print("3. Complete a task") # New option
        print("4. Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            add_task()
        elif choice == '2':
            view_tasks()
        elif choice == '3':
            complete_task()
        elif choice == '4':
            print("Exiting Task Manager. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 4.")

if __name__ == "__main__":
    main()
```
