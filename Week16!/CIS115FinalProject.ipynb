```python
# Expense Tracker #


# IMPORT STATEMENTS #
# Import the datetime module
import datetime
from datetime import datetime

# FUNCTION DEFINITIONS #

# Create a function that filters expenses by categories or vendors with parameters of the dictionary and the filter (key) and its value
def filter_expenses(expense_list, filter_key, filter_value):
  # Create a list to hold filtered expenses
  filtered_list = []
  # Create a "for" loop to find the expense in the expense list dictionary
  for expense in expense_list:
    # First, determine if there are items under the category or vendor
    if expense[filter_key].lower() == filter_value.lower():
      # If it exists, add it to the filtered list
      filtered_list.append(expense)
  # Return the filtered list
  return filtered_list

# Create a function that displays the user's expenses with the expense_list as the parameter
def display_expenses(expense_list):
  # First check if the list has expenses in it
  if not expense_list:
    # If not, print "No expenses to display"
    print("No expenses to display.")
    return
  # if yes, print the expense display header ("Your Expenses:")
  print("Your Expenses:")
  # Create a for loop that goes through each expense (i) and numbers it (enumerate function)
  for i, expense in enumerate(expense_list):
    # Print the list by outlining the category, vendor, amount, and date of each expense, and correct decimal display to two places
    print(f"{i+1}. Category: {expense['category']}, Vendor: {expense['vendor']}, Amount: ${expense['amount']:.2f}, Date: {expense['date']}")

# PROGRAM LOGIC #

# Create a list to store the different expense dictionaries (vendor, category, date, amount)

expenses = []

# Create a while loop that will provide the user with a menu and will break once the user types the number option for "exit"
  # While True:
while True:
    # Print the different options in the menu and create a header: "expense tracker", "add new expenses", "get totals", and "exit"
    print("--- Expense Tracker Menu: ----")
    print(
        "1. Add new expenses\n"
        "2. Specific Total (Vendor/Category)\n"
        "3. Total Expenses\n"
        "4. Exit" )
    # Ask for user input and initialize it as "choice"
    choice = int(input("Please select an option (1-4) to get started: "))
      # if choice = 1
    if choice == 1:
        # Ask user for category, vendor, amount (float), and date
        category = input("Please enter the category of the expense: ")
        vendor = input("Please enter the vendor of the expense: ")
        try:
            amount = float(input("Please enter the amount of the expense: "))
        # To ensure program doesn't crash if a user types a letter and not a number
        except ValueError:
            print("Invalid amount. Please enter a numerical value.")
            continue

        # Get date input as a string and put it into a datetime object
        date_str = input("Please enter the date of the expense (MM/DD/YYYY): ")
        # Assistance from AI with this part
        try:
            date = datetime.strptime(date_str, '%m/%d/%Y').date() # Store only the date part
        except ValueError:
            print("Invalid date format. Please use MM/DD/YYYY.")
            # Skip to the next iteration of the loop
            continue

        # Create a dictionary for the expense being added, separated by category
        current_expense = {
            "category": category,
            "vendor": vendor,
            "amount": amount,
            "date": date
        }
        # Append the expense to the expenses list
        expenses.append(current_expense)

        print(f"Expense added: Category: {category}, Vendor: {vendor}, Amount: ${amount:.2f}, Date: {date}")
      # elif choice = 2
    elif choice == 2:
        # Ask user what total they would like (by category or by vendor); use .upper() to ensure the program recognizes all forms of c and v
        filter_type = input("Would you like to filter by category or vendor? (C/V): ").upper()
        if filter_type == 'C':
            filter_value = input("Enter the category to filter by: ")
            # Call the filter_expenses function and use category (C) as the filter_key
            filtered_expenses = filter_expenses(expenses, 'category', filter_value)
            # Call the display expenses function to show the user the expenses they requested from that category
            display_expenses(filtered_expenses)
            # Sum the amount for all the expenses under that category
            total_filtered = sum(expense['amount'] for expense in filtered_expenses)
            # Display results to two decimal places
            print(f"Total for {filter_value} expenses: ${total_filtered:.2f}")
        elif filter_type == 'V':
            filter_value = input("Enter the vendor to filter by: ")
            # Call the filter_expenses function and replace the filter_key with vendor (v)
            filtered_expenses = filter_expenses(expenses, 'vendor', filter_value)
            # Call the display_expenses function
            display_expenses(filtered_expenses)
            # Sum the total for that vendor
            total_filtered = sum(expense['amount'] for expense in filtered_expenses)
            # Print the results to the user
            print(f"Total for {filter_value} expenses: ${total_filtered:.2f}")
            # If C or V is not entered, send the user an error message
        else:
            print("Invalid filter type. Please enter 'C' for category or 'V' for vendor.")
      # elif choice = 3
    elif choice == 3:
        # total = sum of the amounts of the expenses
        total_expenses = sum(expense['amount'] for expense in expenses)
        # print total to two decimal places
        print(f"Total expenses: ${total_expenses:.2f}")
      # else
    elif choice == 4:
        # Print "Goodbye!"
        print("Goodbye!")
        # Break the loop
        break
        # Send the user an error message if they type a number not 1-4
    else:
        print("Invalid choice. Please select an option between 1 and 4.")
```
