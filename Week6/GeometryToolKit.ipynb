```python


# IMPORT STATEMENTS
# 1. Import Math

import math

# FUNCTION DEFINITIONS
# 1. Define area(radius): returns pi * r squared

def area_of_circle(radius):
  area = (math.pi * (radius**2))
  return area

# 2. Define circumference(radius): returns 2 * pi * r squared

def circumference_of_circle(radius):
  circumference = (math.pi * 2) * (radius)
  return circumference
# 3. Define volume(radius, height):

def volume_of_cylinder(radius, height):
 #   --Call area(radius) and multiply by height
  volume = area_of_circle(radius) * height
 #   --Return the result
  return volume


# PROGRAM LOGIC
# 1. Input for radius (float)

radius = float(input("What is the radius of the circle? (Numbers only): "))

# 2. Call and print area and circumference

area = area_of_circle(radius)
circumference = circumference_of_circle(radius)

print("Area is: " + str(area))
print("Circumference is: " + str(circumference))

# 3. Input for height (float)

height = float(input("What is the height of the cylinder? (Numbers Only): "))

# 4. Call and print volume

volume = volume_of_cylinder(radius, height)
print("Volume is: " + str(volume))
```
