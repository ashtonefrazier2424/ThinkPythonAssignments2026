```python


# Name: gradebook.py

##### IMPORT STATEMENTS #####




##### FUNCTION DEFINITIONS #####

# Define a function called add_students that takes a dictionary ("gradebook") as a parameter
def add_students(gradebook):
  # Create a while loop
  while True:
    # Ask the user for the student's name (name = input)
    name = input("Enter the student's name (or 'done' to finish):")
    # Break when user types "done" and is not case sensitive
    if name.lower() == "done":
      break
    # Ask the user for the student's grade (0-100); (grade = student's grade from 0-100)
    grade = int(input("Enter the student's grade (0-100):"))
    # Make sure number falls within parameters (larger than zero, but less than 100)
    if grade <= 0 or grade <= 100:
    # Add the student's name and grade to the gradebook dictionary
      gradebook[name] = grade
    else:
      print("Invalid number, please try again!")
  # Return gradebook
  return gradebook

# Define a function called calculate_average that takes "gradebook" as a parameter
def calculate_average(gradebook):
  # Avoid division by zero by specifying that if "gradebook" is empty, it should return an average of zero
  if len(gradebook) == 0:
    return 0
  # Calculate the average:
  # Find total by adding all the values from the "gradebook" dictionary using .values method
  total = sum(gradebook.values())
  # Find the average by dividing the total by the number of students using the len function
  average = total/len(gradebook)
  # Return average
  return average


##### PROGRAM LOGIC #####

# Initialize an empty dictionary called "gradebook"

gradebook = {}


# Call the add_students function to fill the dictionary

add_students(gradebook)

# Calculate the class average using the calculate_average function

average = calculate_average(gradebook)

# Print the final report, showing the student's name and corresponding grade
print("Class Report:")
for name, grade in gradebook.items():
  print(f"{name}: {grade}")

# Call the calculate_average function
  # Display the resulting class average to two decimal places
print(f"Class average: {average:.2f}")
```
