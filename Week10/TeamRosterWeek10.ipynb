```python


# NAME: roster.py
# IMPORT STATEMENTS




# FUNCTION DEFINITIONS


# Add players to the list: (Roster)
# Don't have player name as an argument because it's defined in the function
def add_players(team_roster):
  # Create a while loop that asks users to enter a player's name
  # While True:
  while True:
    player_name = input("Enter player name (or 'quit' to stop):").lower()
    # Loop breaks when user types "quit"
    if player_name.lower() == "quit":
      break
  # Add player's name to the roster using the append method (roster.append(player_name))
    team_roster.append(player_name)

# Is on the roster: (Roster, Player Lookup)
def is_on_roster(team_roster, player_lookup):
  # if player name is on the list:
  if player_lookup in team_roster:
    # return True
    return True
  # else:
  else:
    # return false
    return False

# PROGRAM LOGIC

# Create an empty list titled "Roster"

team_roster = []


# Call the add players to the list function(roster)

add_players(team_roster)

# Ask user for input on player they're looking for
# Make sure user's input is not case sensitive

player_lookup = input("Who would you like to search for?").lower()

# Call is on the roster function(roster, player name)

is_on_roster(team_roster, player_lookup)

# result = is on the roster function

result = is_on_roster(team_roster, player_lookup)

# Display findings to user:

# if result:
if result == True:
  # print("Result: Player found!")
  print("Result: Player found!")
# else:
else:
  # print("Result: Player not found!")
  print("Result: Player not found!")
```
